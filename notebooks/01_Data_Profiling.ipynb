```python
print("Passengers > Capacity:", (df["Passengers"] > df["Aircraft_Capacity"]).sum())
print("Load Factor outside 0-1:", ((df["Load_Factor"] < 0) | (df["Load_Factor"] > 1)).sum())
print("Negative Revenue:", (df["Total_Revenue"] < 0).sum())
print("Negative Cost:", (df["Total_Cost"] < 0).sum())
```

```python
profit_check = df["Total_Revenue"] - df["Total_Cost"]

(profit_check - df["Profit"]).abs().describe()
```

```python
margin_check = (df["Profit"] / df["Total_Revenue"]) * 100

(df["Profit_Margin"] - margin_check).abs().describe()
```
